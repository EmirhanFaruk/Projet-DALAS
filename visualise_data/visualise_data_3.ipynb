# IMPORT

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

# DEFINE FOLDERS

```python
DATASET_INPUT_FOLDER = Path("correlated_data_csv")
PLOT_FOLDER = Path("plots3")

folders = [DATASET_INPUT_FOLDER, PLOT_FOLDER]
for folder in folders:
    folder.mkdir(parents=True, exist_ok=True)
```

# LOAD DATA

```python
df = pd.read_csv(f"{DATASET_INPUT_FOLDER}/correlated_data_2.csv")
df.head(5)
```

```python
FEATURES = [
    "numbikesavailable",
    "weather_temperature_2m",
    "weather_wind_speed_10m",
    "weather_wind_direction_10m",
    "weather_weather_code",
    "weather_is_day"
]

for station_id, df_s in df.groupby("stationcode"):
    df_s = df_s.copy()
    df_s["date"] = df_s["weather_time"].dt.date
    df_s["hour"] = df_s["weather_time"].dt.hour

    dates = sorted(df_s["date"].unique())
    if len(dates) < 3:
        continue  # skip tiny stations

    for feature in FEATURES:
        presence = np.zeros((len(dates), 24))  # rows = days, cols = hours

        for i, d in enumerate(dates):
            hours_present = df_s[df_s["date"] == d].loc[df_s[feature].notna(), "hour"].unique()
            presence[i, hours_present] = 1

        # Plot
        plt.figure(figsize=(14, 6))
        plt.imshow(presence, aspect="auto", cmap="Greys", vmin=0, vmax=1)
        plt.xticks(range(24), range(24))
        plt.yticks([])  # too many days to label
        plt.xlabel("Hour of day")
        plt.title(f"Station {station_id} — Feature: {feature} — Missing data pattern")
        plt.colorbar(label="Data present (1) / Missing (0)")
        plt.tight_layout()
        plt.savefig(f"{PLOT_FOLDER}/station_{station_id}_{feature}.png")
        plt.close()

print("Feature-missingness plots saved.")
```
