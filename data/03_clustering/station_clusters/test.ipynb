```python
import pandas as pd
```

```python
def compare_clusters(file_behavior, file_behavior_weather):
    behavior = pd.read_csv(file_behavior)
    behavior_weather = pd.read_csv(file_behavior_weather)

    merged = behavior.merge(behavior_weather, on="station_code", suffixes=("_behavior", "_weather"))

    merged["changed"] = merged["cluster_behavior"] != merged["cluster_weather"]

    num_changed = merged["changed"].sum()
    total_stations = merged.shape[0]
    percent_changed = num_changed / total_stations * 100

    print(f"{file_behavior}    VS    {file_behavior_weather}")
    print(f"{num_changed} out of {total_stations} stations changed cluster ({percent_changed:.2f}%)")

```

```python
compare_clusters(
    "station_clusters_kmeans_k2_feat_behavior_rs42_old.csv",
    "station_clusters_kmeans_k2_feat_behavior+weather_rs42_old.csv"
)

compare_clusters(
    "station_clusters_kmeans_k2_feat_behavior_rs42.csv",
    "station_clusters_kmeans_k2_feat_behavior+weather_rs42.csv"
)

compare_clusters(
    "station_clusters_kmeans_k2_feat_behavior_rs42.csv",  
    "station_clusters_kmeans_k2_feat_behavior_rs42_old.csv"
)

compare_clusters(
    "station_clusters_kmeans_k2_feat_behavior+weather_rs42.csv",
    "station_clusters_kmeans_k2_feat_behavior+weather_rs42_old.csv"
)
```
